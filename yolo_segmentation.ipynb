```python
from ultralytics import YOLO

class YOLOTrainer:
  def __init__(self)-> None:
    self.model = YOLO("yolo11m-seg.pt")

  def train_model(self):
    train_results = self.model.train(data="/content/data.yaml",
                                     epochs=30,
                                     imgsz=640,
                                     batch=16,
                                     warmup_epochs=3,
                                     fliplr=0.1,
                                     mosaic=0.4,
                                     close_mosaic=10,
                                     hsv_h=0.08,
                                     hsv_s=0.06,
                                     hsv_v=0.07,
                                     copy_paste=0.15,
                                     scale=0.2,
                                     weight_decay = 0.00005)

    return train_results

  def val_model(self):
    val_model = self.model.val(data="/content/data.yaml")
    return val_model

  def predict_model(self):
    results = self.model.predict(source="/content/test/images",
                                 save=True,
                                 conf=0.2)

    return results

  def export_model(self):
      self.model.export(format="onnx")


model = YOLOTrainer()
train_results = model.train_model()
val_results = model.val_model()

print (f"Train Results {train_results}")
print (f"Validation Results {val_results}")

results = model.predict_model()
for result in results:
  boxes = result.boxes

  if boxes is not None:
    for box in boxes:
      confidence = float(box.conf[0])
      class_id = int(boxes.cls[0])
      class_name = result.names[class_id]

      print (f"Confidence {confidence}")
      print (f"Class_ID {class_id}")
      print (f"Class_Name {class_name}")

model.export_model()
```
